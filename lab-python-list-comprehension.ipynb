In the previous exercise, you developed a program to manage customer orders and inventory. Now, let's take it a step further and incorporate comprehension into your code.

    - *Hint: Apply it to initialize inventory, updating the inventory and printing the updated inventory.*
    
    - For example, in initializing the inventory, we could have:

        def initialize_inventory(products):

````python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
````

````python
def initialize_inventory(products):
    """Create an inventory using dictionary comprehension."""
    inventory = {
        product: int(input(f"Enter the quantity of {product}s available: "))
        for product in products
    }
    return inventory
````

````python
inventory = initialize_inventory(products)
inventory
````

````python
inventory
````

2. Modify the function get_customer_orders so it prompts the user to enter the number of customer orders and gathers the product names using a loop and user input. Use comprehension.

````python
def get_customer_orders(products):
    """Collect the requested products using a set comprehension."""
    items_to_order = int(input("How many items do you want to order?: "))

    customer_orders = {
        input("Please add one product name you want to order: ").strip().lower()
        for _ in range(items_to_order)
    }

    # Keep only products that exist in the available product list.
    customer_orders = {
        product for product in customer_orders
        if product in products
    }

    return customer_orders
````

````python
customer_orders = get_customer_orders(products)
customer_orders
````

4. Add a new function to calculate the total price of the customer order. For each product in customer_orders, prompt the user to enter the price of that product. Use comprehension to calculate the total price. Note: assume that the user can only have 1 unit of each product.

````python
def calculate_order_total(customer_orders):
    """Ask for each product price and return the total order price."""
    item_prices = [
        float(input(f"Enter the price of {product}: "))
        for product in customer_orders
    ]

    return sum(item_prices)
````

````python
total_order = calculate_order_total(customer_orders)
print(f"Total Price: {total_order}")
````

4. Modify the update_inventory function to remove the product from the inventory if its quantity becomes zero after fulfilling the customer orders. Use comprehension to filter out the products with a quantity of zero from the inventory.

````python
def update_inventory(inventory, customer_orders):
    """Update quantities and remove products that reach zero."""
    updated_inventory = {
        product: quantity - 1 if product in customer_orders else quantity
        for product, quantity in inventory.items()
    }

    updated_inventory = {
        product: quantity
        for product, quantity in updated_inventory.items()
        if quantity > 0
    }

    return updated_inventory
````

````python
def print_updated_inventory(inventory):
    """Print the updated inventory in a readable format."""
    messages = {
        product: f"The updated amount of {product}s is: {quantity}"
        for product, quantity in inventory.items()
    }

    for message in messages.values():
        print(message)
````

Final program execution
The functions are called in sequence so the complete customer-order workflow can be executed.

````python
updated_inventory = update_inventory(inventory, customer_orders)
print_updated_inventory(updated_inventory)
````
